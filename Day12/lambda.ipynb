```
# Lambda function example
#lambda examples// Anonymous function


# Q1.wap to find square and cube of given number

res=lambda x:x*x
# res(5)
print(res(3))
```

```
# 2.wap to check the given number is palindrome or not
txt=input('enter txt: ')
res=lambda x:x==x[::-1]
if res(txt):
  print('plaindrome')
else:
  print('not a plaindrome')
```

```
txt=input('enter txt: ')
print('plaindrom' if (lambda x:x==x[::-1].lower())(txt) else 'not plaindrom' )
```

```
# 3.wap to convert negative number to positive number
no =int(input('enter no: '))
nev= lambda x:-x if x<0 else x
print(nev(no))
```

```
# 4.wap to return the key of a dictionary
a={"hello":"Sony","How":"are","you":"bye"}
ab=lambda x:x.keys()
print(list(ab(a)))
```

```
# 5.wap to check if the number is even or odd
no =int(input('enter no: '))
c=lambda x: 'even'if x%2==0 else'not even'
print(c(no))
```

```
# 6.wap which returns first element of a sequence
seq = [10, 20, 30, 40]
first = lambda x: x[0]
print(first(seq))
```

```

# 7.wap which returns length of any iterable
data = "python"
length = lambda x: len(x)
print(length(data))
```

```
# 8.wap which returns the list of squares of
# numbers in a list
l=[2,3,4,5,6]
no=(map(lambda x:x**2,l))
print(list(no))
```

```
# 10.wap to print the numbers present in a list
# with their corresponding indices pair
l=[100,200,300,400,500]
no=map(lambda x: (x[0],x[1]),enumerate(l))
print(list(no))
```

```


# 11.wap to check a data is sequence if it is
# sequence then return length pair else type pair
data = [10, 20, 30,'k',(8)]

check = lambda x: ("Length", len(x)) if isinstance(x, (list, tuple, str, set, dict)) else ("Type", type(x))

print(check(data))
```

```
# 12.wap to check given number is divisible by 5
# and 3
no = int(input("enter no: "))

div = lambda x: x % 5 == 0 and x % 3 == 0
print(div(no))
```

```
# 13.wap to check even or odd,if it is even return
# square of that value else square root of that
# value
import math

no = int(input("enter no: "))

res = lambda x: x**2 if x % 2 == 0 else math.sqrt(x)
print(res(no))
```

```
# 14.wap to check len of given string ,if length is
# even return as it is else return reverse
# # of that string
s = input("enter string: ")

res = lambda x: x if len(x) % 2 == 0 else x[::-1]
print(res(s))
```
