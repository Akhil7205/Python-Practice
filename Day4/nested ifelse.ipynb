```
# 8.wap to give 10% off only who is purchasing in
# credit card and min 3 product should purchase and
# each product price should be more than 500
payment = input("Enter payment method (credit/debit/cash): ").lower()
count = int(input("Enter number of products: "))

prices = []
for i in range(count):
    price = int(input(f"Enter price of product {i+1}: "))
    prices.append(price)

if payment == "credit" and count >= 3 and all(p > 500 for p in prices):
    total = sum(prices)
    discount = total * 0.10
    final_amount = total - discount

    print("Total Amount:", total)
    print("Discount Applied: 10%")
    print("Final Amount:", final_amount)
else:
    print(" Discount not applicable")
    print("Total Amount:", sum(prices))

```

```

# 9.wap to perform list operations user should
# enter only list data type,if options 1
# pop().options 2 sort() options 3 clear() invalid
# options,invalid data type

user=eval(input('enter list: '))
if type(user)==list:
  print('option avaliables are:1-pop(),2-sort(),3-clear()')
  option=int(input('enter option: '))
  if option==1:
    user.pop()
    print(user)
  elif option==2:
    user.sort()
    print(user)
  elif option==3:
    user.clear()
    print(user)
else:
  print(' invalid datatype ')


```
