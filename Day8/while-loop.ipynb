```
#15.wap to iterate inside the list check if it is having nested list if yes merge it
# list1=["hello",10,20.55,True,False,"hai","bye",[False,"goodnight","enjoy the holiday"]]
# #excepted
# output:-->list1=["hello",10,20.55,True,False,"hai","bye",False,"goo
# dnight","enjoy the holiday"]
list1=["hello",10,20.55,True,False,"hai","bye",[False,"goodnight","enjoy the holiday"]]
i=0
d=[]
while i<len(list1):
  if isinstance(list1[i],list):
    for j in list1[i]:
      d.append(j)
  else:
    d.append(list1[i])
  i+=1
print(d)
```
