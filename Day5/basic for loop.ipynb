```
# 2.wap to extract vowels and digits in a string
s="hello123"
vowels=['a','e','i','o','u']
v=[]
d=[]
for i,j in enumerate(s):
  if j.lower() in vowels:
    v.append((i,j))
  elif j.isdigit():
    d.append((i,j))
print(v)
print(d)
```

```
# 6.wap to print the count of alphabets and numbers
# and space in the given string
t="india got the independence in the year 1947"
s=[]
a=[]
d=[]
for i in t:
  if i==" ":
    s+=i
  elif i.isalpha():
    a+=i
  elif i.isdigit():
    d+=i
print('space',len(s))
print('alpha',len(a))
print('digit',len(d))
```
