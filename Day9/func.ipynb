```
# 11.wap to check given iterable is a sequence,if
# it is a sequence reverse it,if not add one extra
# element to the iterable
def seq(rev):
  if isinstance(rev,(list,tuple,str)):
    return rev[::-1]
  else:
   rev.add('extra')
   return rev

# seq('asdf')
s={2,2,2}
print(seq(s))
```
