```
# 1) WAP to generate a+b, a-b, a*b, a/b by taking a and b from user
def operations():
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))

    print("a+b =", a + b)
    print("a-b =", a - b)
    print("a*b =", a * b)

    if b != 0:
        print("a/b =", a / b)
    else:
        print("Division not possible (b=0)")

operations()
```

```
# 2) WAP to generate only values divisible by 5
def divisible_by_5(l):
    return [i for i in l if i % 5 == 0]

l=[34,55,60,56,78,90,25,40]
print(divisible_by_5(l))
```

```
# 3) WAP to return an iterator having square root of list values
import math

def sqrt_iterator(l):
    return map(math.sqrt, l)

l=[25,36,49,81,9,16]
print(list(sqrt_iterator(l)))
```

```
# 4) WAP to return iterator of (word, len(word)) and typecast into dictionary
def word_len_dict(l):
    return dict(map(lambda x: (x, len(x)), l))

l=["instagram","facebook","whatsapp","meta","oracle"]
print(word_len_dict(l))
```

```
# 5) WAP to generate only numeric values in given list
def only_numeric(l):
    return [i for i in l if type(i) in (int, float, complex)]

l=["flipkart","Amazon",78,[2,3,4],78,9.87,(5,3),45.36]
print(only_numeric(l))
```

```
# 6) WAP to generate list: if individual datatype reverse it else keep same
def reverse_individual(l):
    res = []
    for i in l:
        if type(i) == str:
            res.append(i[::-1])
        elif type(i) in (int, float):
            res.append(str(i)[::-1])
        else:
            res.append(i)
    return res

l=["flipkart","Amazon",78,[2,3,4],78,9.87,(5,3),45.36]
print(reverse_individual(l))
```

```
# 7) WAP to generate only strings with odd length
def odd_length_strings(l):
    return [i for i in l if len(i) % 2 != 0]

l=["alexa","siri","google","cortrena"]
print(odd_length_strings(l))
```

```
# 8) WAP even → square, odd → cube
def even_square_odd_cube(l):
    return [i**2 if i%2==0 else i**3 for i in l]

l=[2,3,4,5,6,7]
print(even_square_odd_cube(l))
```

```
# 9) WAP reverse words if even length
def reverse_even_length_words(l):
    return [i[::-1] if len(i)%2==0 else i for i in l]

l=["hello","world","python","apple","google","walmart"]
print(reverse_even_length_words(l))
```

```
# 10) WAP first letter as key and words as values (dictionary)
def group_by_first_letter(s):
    d = {}
    words = s.split()

    for w in words:
        key = w[0]
        d.setdefault(key, []).append(w)

    return [d]

s="python is a programming language and programming is part of life"
print(group_by_first_letter(s))
```

```
# 11) WAP reverse only individual datatype else keep it as it is
def reverse_single_datatypes(a):
    res = []
    for i in a:
        if type(i) == str:
            res.append(i[::-1])
        elif type(i) in (int, float, complex, bool):
            res.append(str(i)[::-1])
        else:
            res.append(i)
    return res

a=["good",45,[1,2],78.6,(4,5),8+7j,{9,7},False,{"a":75}]
print(reverse_single_datatypes(a))
```
