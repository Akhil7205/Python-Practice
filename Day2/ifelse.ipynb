```python
# 17.wap to check the given number is odd, if it is odd divide it by 2 and print reminder and quotient else print it is even (take userinput)

num = int(input("Enter a number: "))

if num % 2 != 0:
    quotient = num // 2
    remainder = num % 2
    print("Odd number")
    print("Quotient:", quotient)
    print("Remainder:", remainder)
else:
    print("It is even")
```

```python
# 19.WAP to check whether the given number value is divisible by 6 or
# not,if it is divisible cube that number or else perform left shift operation by 3
# (take user input)
num = int(input("Enter a number: "))

if num % 6 == 0:
    print("Divisible by 6")
    print("Cube:", num ** 3)
else:
    print("Not divisible by 6")
    print("Left shift by 3:", num << 3)
```
