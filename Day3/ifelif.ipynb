```
# elif Condition Programs:
# 1.wap to check whether the given character is uppercase/lowercase/digit/special (with and without using inbuilt function
ch = input("Enter a single character: ")

if ch.isupper():
    print("Uppercase letter")
elif ch.islower():
    print("Lowercase letter")
elif ch.isdigit():
    print("Digit")
else:
    print("Special character")
```

```
ch = input("Enter a single character: ")

ascii_val = ord(ch)

if ascii_val >= 65 and ascii_val <= 90:
    print("Uppercase letter")
elif ascii_val >= 97 and ascii_val <= 122:
    print("Lowercase letter")
elif ascii_val >= 48 and ascii_val <= 57:
    print("Digit")
else:
    print("Special character")
```

```
# 2.wap to check a data is a sequence/iterable/individual data type
import ast
data=input('enter any data: ')
data=ast.literal_eval(data)

if isinstance(data, (str, list, tuple)):
    print("Sequence data type")
elif isinstance(data, (set, dict)):
    print("Iterable data type")
elif isinstance(data, (int, float, complex, bool)):
    print("Individual data type")
else:
    print("Unknown data type")
```

```
# 7.wap to take marks of 5 sub,calculate the average if
# the average is b/w 90-100 print Distinction
# if 75-89 print first class and if it's 60-74 print
# second class, if 50-59 print Third class,below 50 is
# fail
# note:-->max marks is 100
m1 = int(input("Enter marks of subject 1: "))
m2 = int(input("Enter marks of subject 2: "))
m3 = int(input("Enter marks of subject 3: "))
m4 = int(input("Enter marks of subject 4: "))
m5 = int(input("Enter marks of subject 5: "))

# Calculate average
avg = (m1 + m2 + m3 + m4 + m5) / 5

print("Average:", avg)

if avg >= 90 and avg <= 100:
    print("Distinction")
elif avg >= 75 and avg <= 89:
    print("First Class")
elif avg >= 60 and avg <= 74:
    print("Second Class")
elif avg >= 50 and avg <= 59:
    print("Third Class")
else:
    print("Fail")
```

```
# 10.wap to give discount to customer based on total
# price(p1+p2+p3) 1000 to 3000 price 500 discount and
# 3001 to 5000 price 1000 discount more than 5001 price
# 1200 discount and less than 1000 price no discount.

p1 = int(input("Enter price of product 1: "))
p2 = int(input("Enter price of product 2: "))
p3 = int(input("Enter price of product 3: "))

total = p1 + p2 + p3
print("Total price:", total)

if 1000 <= total <= 3000:
    discount = 500
elif 3001 <= total <= 5000:
    discount = 1000
elif total >= 5001:
    discount = 1200
else:
    discount = 0

final_price = total - discount

print("Discount:", discount)
print("Final amount to pay:", final_price)
```
