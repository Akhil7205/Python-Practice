```
l = [[1,2,3], [4,5,6], [7,8,9]]

external = 0
internal = []

for i in l:
    temp_sum = 0
    for j in i:
        temp_sum += j       # internal sum
        external += j       # external sum
    internal.append(temp_sum)

print("internal:", internal)
print("external:", external)
```

```
# 13.WAP to print sum of internal and external list
# l = [[1,2,3], [4,5,6], [7,8,9]]
# output:-->
#internal = 6, 15, 24
#external--> 45

l = [[1,2,3], [4,5,6], [7,8,9]]
external=0
internal=[]
for i in l:
  sum=0
  for j in i:
      sum+=j
      external+=j
  internal.append(sum)
print('external:',external)
print('internal',internal)
```

```

# 14.wap to using this list get the below output
# l = ['sun flower', 'Lilly flower', 'Marigold
# flower', 'lion animal', 'tiger animal', 'eagle
# bird', 'snake animal', 'lotus flower', 'pigeon
# bird']
# o/p:-->{'flower': ['sun', 'Lilly', 'Marigold',
# 'lotus'], 'animal': ['lion', 'tiger', 'snake'],
# 'bird': ['eagle', 'pigeon']}
l = ['sun flower', 'Lilly flower', 'Marigold flower', 'lion animal', 'tiger animal', 'eagle bird', 'snake animal', 'lotus flower', 'pigeon bird']
d={}
for i in l:
  value,key=(i.split())
  if key not in d:
    d[key]=[]
  d[key].append(value)
print(d)
```
